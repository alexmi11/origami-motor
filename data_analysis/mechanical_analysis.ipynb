```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from itertools import product
import math
import scienceplots

plt.style.use(['science','nature'])
plt.rcParams.update({'figure.dpi': '300'})

from scipy import signal
from scipy import stats
from scipy.interpolate import make_smoothing_spline

COLLAPSED_SIZE = 16 # mm
DEPLOYED_SIZE = 62 # mm
ROTOR_POSITION_OFFSET = 8 # mm
rotor_index_offset = -1


df_mech = pd.read_csv('20260908_mechanical_data/force_displacement.csv', delimiter=",")
#print(df_mech)


plt.plot(df_mech["L_total(mm)"], df_mech["force_total(N)"], label='Full assembly')
plt.plot(df_mech["L_stator(mm)"], df_mech["force_stator(N)"], label='Stator')
plt.plot(df_mech["L_rotor(mm)"][0:rotor_index_offset]+ROTOR_POSITION_OFFSET, df_mech["force_rotor(N)"][0:rotor_index_offset], label='Rotor')
plt.plot(df_mech["L_rotor(mm)"][rotor_index_offset-1:]+ROTOR_POSITION_OFFSET, df_mech["force_rotor(N)"][rotor_index_offset-1:], color = "orange", alpha = 0.5, linestyle="dashed")


plt.plot(np.nan, np.nan, '-', color='none', label=' ') # fake plot to take space in the legend.

plt.axvline(COLLAPSED_SIZE, color = 'grey', alpha = 0.7, linestyle = "dotted", label = "Collapsed")
plt.axvline(DEPLOYED_SIZE, color = 'grey', alpha = 0.7, linestyle = "dashdot", label = "Deployed")


plt.xlim(0,63.5)
plt.ylim(0,)

plt.xlabel("Motor length (mm)")
plt.ylabel("Force (N)")

leg = plt.legend(loc='upper right')

plt.savefig("figures_20260817/mechanical.pdf", format="pdf")  
```

```python
#print(df_mech)
k_rotor_2_mm = -df_mech["force_rotor(N)"][1] / (1e-3 * (df_mech["L_rotor(mm)"][1] - df_mech["L_rotor(mm)"][0]))

print("k_rotor_2mm (N/m): ")
print(k_rotor_2_mm)

M_rotor = 8.9 * 1e-3 # kg.

f_rotor_resonant = (1/ (2 * np.pi)) * np.sqrt(k_rotor_2_mm / M_rotor)

print("Resonant frequency of rotor structure at small deformations: (Hz): ")
print(f_rotor_resonant)

print("Corresponding excitation speed (RPM): ")
print(f_rotor_resonant * 60)
```

```python

k_stator_2_mm = -df_mech["force_stator(N)"][1] / (1e-3 * (df_mech["L_stator(mm)"][1] - df_mech["L_stator(mm)"][0]))

print("k_stator_2mm (N/m): ")
print(k_stator_2_mm)

M_stator = 31 * 1e-3 # kg

f_stator_resonant = (1/ (2 * np.pi)) * np.sqrt(k_stator_2_mm / M_stator)

print("Resonant frequency of stator structure at small deformations: (Hz): ")
print(f_stator_resonant)

print("Corresponding excitation speed (RPM): ")
print(f_stator_resonant * 60)
```
