```python
from sympy import *
import numpy as np

init_printing(use_unicode=True)

epsilon_1 = symbols('epsilon_1', real=True, positive=True)
epsilon_2 = symbols('epsilon_2', real=True, positive=True)
sigma_1 = symbols('sigma_1', real=True, positive=True)
sigma_2 = symbols('sigma_2', real=True, positive=True)

R = symbols('R', real=True, positive=True)
L = symbols('L', real=True, positive=True)
G = symbols('G', real=True, positive=True)
omega = symbols('omega', real=True)

theta = symbols('theta', real=True)
alpha = symbols('alpha', real=True)
tau = symbols('tau', real=True)

J = symbols('J', real=True)

sigma_1 = (epsilon_1 + epsilon_2) / tau

E_0 = symbols('E_0', real=True)

A =  R**2 * ( (-E_0 *   ((sigma_1+ I * omega * (epsilon_2 - epsilon_1)) ) -  J   * exp(I*(-alpha)) ) / (sigma_1 + I * omega * (epsilon_2 + epsilon_1)))
B = E_0 - (A / R**2)

rho_fs = re((epsilon_2 * (E_0 + A/R**2) - epsilon_1*B)* exp(I*(theta)) )
print("rho_fs: \n \n \n")
pprint(rho_fs.simplify().simplify())
pprint(latex(rho_fs.simplify()))

E_a_r = re((E_0  + (A / (R**2))) * exp(I*(theta)) )
E_a_theta = re(I*(E_0  - (A / (R**2))) * exp(I*theta) )


print("\n \n Torque, calculated with stress tensor: \n \n \n")
integrand = epsilon_2 * R*  E_a_theta * E_a_r  * R * L
torque = integrate(integrand, (theta, 0 , 2*pi))
torque_simplified = factor(torque.simplify())
pprint(latex(torque_simplified))
pprint((torque_simplified))


print("\n \n Torque, calculated with rho_fs: \n \n \n")
integrand2 = R *  E_a_theta * rho_fs  * L * R
torque2 = integrate(integrand2, (theta, 0 , 2*pi))
torque_simplified2 = factor(torque2.simplify())
pprint(latex(torque_simplified2))
pprint((torque_simplified2))
```
